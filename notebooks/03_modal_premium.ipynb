```python
import pandas as pd
```

```python
modal_factors = {
    "Annual": 1.000,
    "Half-yearly": 0.507,
    "Quarterly": 0.255,
    "Monthly": 0.086
}

modal_factors["Quarterly"]
```

```python
modal_df = pd.DataFrame(
    {
        "Mode": list(modal_factors.keys()),
        "Factor": list(modal_factors.values()),
        "Payments per year": [1, 2, 4, 12],
    }
)
modal_df
```

```python
def modal_premium(annual_premium, mode):
    """Return the instalment premium for a payment mode"""
    factor = modal_factors[mode]
    return round(annual_premium * factor, 2)
```

```python
print(modal_premium(100000, "Monthly"))
modal_premium(100000, "Quarterly")
```

```python
annual_premium = 100000

modal_df["Instalment premium"] = modal_df["Mode"].apply(
    lambda m: modal_premium(annual_premium, m)
)
modal_df["Total paid in a year"] = (
    modal_df["Instalment premium"] * modal_df["Payments per year"]
)
modal_df
```

```python
assert modal_premium(100000, "Annual") == 100000
print("Check passed")
```
